### Teste com documento

```python
base_url = "http://127.0.0.1:5000"
# Upload de arquivos e configuração da cadeia QA

upload_url = f"{base_url}/upload"

with open("configs/config_Titulo_eleitor.json", "r", encoding="utf-8") as file:

    data = json.load(file)
```

```python
data["path_selected_filename"] = "docs/test_documents/Título_frente.jpg"

data["message_template"] = (
    "Não forneça uma resposta que não está de acordo com o contexto. Perguntas serão separadas por //.Seja direto. Responda somente o necessário. Qualquer data deve ser fornecida no formato MM-DD-YYYY. A respota dever ser um json onde key é a pargunta e value a respota"

)
data[
    "context_template"
] = r"""
        {context}
        O número de inscrição tem 12 dígitos.
        O DV tem dois digitos, os dois últimos digitos do número de incrição de 12 dígitos
        Zona tem 3 dígitos.
        Seção tem 4 dígitos.
        """
response = requests.post(upload_url, json=data)

print("Upload Response:", response.json())
```

```python
# Enviar uma pergunta
chat_url = f"{base_url}/chat"
question = {
    "message": "Tipo de Documento//Nome do Eleitor//Data de nascimento//Número de inscrição//DV//Zona//Seção//Data de Emissão//"
}
response = requests.post(chat_url, json=question)
print("Chat Response:", response.json())
```
